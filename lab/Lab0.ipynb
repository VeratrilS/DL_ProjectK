```python
import numpy as np

n = int(input("Enter the order of the square matrix: "))
print("Enter the matrix elements row-wise:")

A = []
for i in range(n):
    row = list(map(int, input().split()))
    A.append(row)

A = np.array(A)

print("\nMatrix A:")
print(A)

# Manual transpose
AT = []
for i in range(n):
    row = []
    for j in range(n):
        row.append(A[j][i])
    AT.append(row)

AT = np.array(AT)

print("\nTranspose of A:")
print(AT)

product = np.dot(A, AT)

print("\nA x Aᵀ =")
print(product)

I = np.eye(n, dtype=int)

if np.array_equal(product, I):
    print("\nThe matrix is Orthogonal.")
else:
    print("\nThe matrix is NOT Orthogonal.")
```

```python
import numpy as np

print("Enter the elements of the 3x3 matrix row-wise:")

A = []
for i in range(3):
    row = list(map(float, input().split()))
    A.append(row)

A = np.array(A)

print("\nMatrix A:")
print(A)

det = np.linalg.det(A)

if det != 0:
    print("\nThe matrix is Invertible.")
    inverse = np.linalg.inv(A)
    print("\nInverse of the matrix:")
    print(inverse)
else:
    print("\nThe matrix is NOT Invertible.")
```

```python
import numpy as np

n = int(input("Enter the order of the square matrix: "))
print("Enter the matrix elements row-wise:")

A = []

for i in range(n):
    row = list(map(int, input().split()))
    A.append(row)

A = np.array(A)

print("\nMatrix A:")
print(A)

trace = 0

for i in range(n):
    trace += A[i][i]

print("\nTrace of the matrix =", trace)
```

```python
import numpy as np

print("Enter the elements of the 4x4 matrix row-wise:")

A = []

for i in range(4):
    row = list(map(float, input().split()))
    A.append(row)

A = np.array(A)

print("\nOriginal Matrix:")
print(A)

U = A.copy()

for i in range(4):
    if U[i][i] == 0:
        print("Pivot is zero. Row swapping required.")
        break

    for j in range(i + 1, 4):
        factor = U[j][i] / U[i][i]

        for k in range(4):
            U[j][k] = U[j][k] - factor * U[i][k]

print("\nUpper Triangular Matrix:")
print(U)

det = 1

for i in range(4):
    det *= U[i][i]

print("\nDeterminant =", det)
```

```python
import numpy as np
# auto orthogonal matrix
# n = np.random.randint(3, 6)
# mat = np.random.rand(n, n)
n = int(input("Enter the size of the matrix: "))
mat = np.zeros((n, n))
for i in range(0,n):
    for j in range(0,n):
        mat[i][j] = int(input(f"Enter element [{i}][{j}]: "))
print(mat)
print(mat.T)
mat = mat @ mat.T
if(np.allclose(mat @ mat.T, np.eye(n))):
    print("The matrix is orthogonal.")
else:
    print("The matrix is not orthogonal.")
```
