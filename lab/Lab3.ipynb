Q1. Implement Batch normalization for the following ANN without using standard library
No. of layers = 4
No. of neurons/layer (excluding output layer) = 3 
No. of neurons in output layer = 2

```python
import numpy as np

def batch_norm(x, gamma, beta, eps):
    mean = np.mean(x, axis=0)
    var = np.var(x, axis=0)
    x_norm = (x - mean) / np.sqrt(var + eps)
    return gamma * x_norm + beta
def relu(x):
    return np.maximum(0, x)

X = np.array([
    [2.3, -1.9, 7.2],
    [2.3, -1.9, 7.2],
    [2.3, -1.9, 7.2]
])

W_ih1 = np.array([-0.2, 1.2, 3.4, -4.3, 2.25, 9.1, -8.9, 0.08, 1.2]).reshape(3, 3)
W_h1h2 = np.array([-0.25, 1.12, 13.4, -4.3, 2.25, 9.12, -9.9, 0.08, 1.25]).reshape(3, 3)
W_h2O = np.array([-0.25, 1.12, -11.4, 4.13, 2.25, 9.12]).reshape(3, 2)

eps = 0.01
lam = 0.2
beta = 1.3

z1 = np.dot(X, W_ih1)
z1_bn = batch_norm(z1, lam, beta, eps)
a1 = relu(z1_bn)

z2 = np.dot(a1, W_h1h2)
z2_bn = batch_norm(z2, lam, beta, eps)
a2 = relu(z2_bn)

z3 = np.dot(a2, W_h2O)
z3_bn = batch_norm(z3, lam, beta, eps)
out = relu(z3_bn)

print(out)
```

Q2. Implement Batch normalization for the following ANN using standard library
No. of layers = 4
No. of neurons/layer (excluding output layer) = 3 
No. of neurons in output layer = 2

```python
import tensorflow as tf
import numpy as np

X = np.array([
    [2.3, -1.9, 7.2],
    [2.3, -1.9, 7.2],
    [2.3, -1.9, 7.2]
])

W_ih1 = np.array([-0.2, 1.2, 3.4, -4.3, 2.25, 9.1, -8.9, 0.08, 1.2]).reshape(3, 3)
W_h1h2 = np.array([-0.25, 1.12, 13.4, -4.3, 2.25, 9.12, -9.9, 0.08, 1.25]).reshape(3, 3)
W_h2O = np.array([-0.25, 1.12, -11.4, 4.13, 2.25, 9.12]).reshape(3, 2)

model = tf.keras.Sequential([
    tf.keras.Input(shape=(3,)),
    tf.keras.layers.Dense(3, use_bias=False),
    tf.keras.layers.BatchNormalization(epsilon=0.01, gamma_initializer=tf.keras.initializers.Constant(0.2), beta_initializer=tf.keras.initializers.Constant(1.3)),
    tf.keras.layers.Activation('relu'),
    tf.keras.layers.Dense(3, use_bias=False),
    tf.keras.layers.BatchNormalization(epsilon=0.01, gamma_initializer=tf.keras.initializers.Constant(0.2), beta_initializer=tf.keras.initializers.Constant(1.3)),
    tf.keras.layers.Activation('relu'),
    tf.keras.layers.Dense(2, use_bias=False),
    tf.keras.layers.BatchNormalization(epsilon=0.01, gamma_initializer=tf.keras.initializers.Constant(0.2), beta_initializer=tf.keras.initializers.Constant(1.3)),
    tf.keras.layers.Activation('relu')
])

model.layers[0].set_weights([W_ih1])
model.layers[3].set_weights([W_h1h2])
model.layers[6].set_weights([W_h2O])

X_tf = tf.constant(X, dtype=tf.float32)
out_tf = model(X_tf, training=True)
print(out_tf.numpy())
```
