```python
# Addition and subtraction using TensorFlow

import tensorflow as tf

a = tf.constant(10)
b = tf.constant(5)

print("Addition =", tf.add(a, b).numpy())
print("Subtraction =", tf.subtract(a, b).numpy())
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

model = Sequential([
    Input(shape=(2,)),
    Dense(2, activation="relu"),
    Dense(2, activation="softmax")
])

model.compile(optimizer="adam",
              loss="categorical_crossentropy",
              metrics=["accuracy"])

model.summary()
```

```python
# ANN (2 input neurons, 2 hidden neurons, 2 output neurons)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential([
    Dense(2, activation="relu", input_shape=(2,)),
    Dense(2, activation="softmax")
])

model.compile(optimizer="adam",
              loss="categorical_crossentropy",
              metrics=["accuracy"])

model.summary()
```
