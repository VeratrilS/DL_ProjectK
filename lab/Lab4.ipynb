## Lab 4: Implement AND, OR, NOR using perceptron

**Instructions:**
a) Assume your weights, bias accordingly.
b) Implement the above in such a way that for a given (w1, w2) the output should be -w0 such that the truth table is satisfied.

### Part A: Assume weights and bias accordingly

```python
def perceptron(x1, x2, w1, w2, w0):
    return 1 if (w1 * x1 + w2 * x2 + w0) >= 0 else 0

def test_gate(gate_name, w1, w2, w0, truth_table):
    print(f'--- {gate_name} Gate (w1={w1}, w2={w2}, w0={w0}) ---')
    inputs = [(0,0), (0,1), (1,0), (1,1)]
    for (x1, x2), expected in zip(inputs, truth_table):
        out = perceptron(x1, x2, w1, w2, w0)
        print(f'  Input: ({x1}, {x2}) -> Output: {out} (Expected: {expected})')
    print()

truth_tables = {
    'AND': [0, 0, 0, 1],
    'OR': [0, 1, 1, 1],
    'NOR': [1, 0, 0, 0]
}

# Assuming weights and bias for each gate
test_gate('AND', w1=1, w2=1, w0=-1.5, truth_table=truth_tables['AND'])
test_gate('OR', w1=1, w2=1, w0=-0.5, truth_table=truth_tables['OR'])
test_gate('NOR', w1=-1, w2=-1, w0=0, truth_table=truth_tables['NOR'])
```

### Part B: For a given (w1, w2) output -w0

```python
def get_minus_w0(w1, w2, gate_name, truth_table):
    print(f'--- {gate_name} Gate ---')
    lower_bound = -float('inf')
    upper_bound = float('inf')
    
    inputs = [(0,0), (0,1), (1,0), (1,1)]
    for (x1, x2), y in zip(inputs, truth_table):
        val = w1*x1 + w2*x2
        if y == 1:
            upper_bound = min(upper_bound, val)
        else:
            lower_bound = max(lower_bound, val)
            
    if lower_bound < upper_bound:
        minus_w0 = upper_bound if upper_bound != float('inf') else lower_bound + 1
        print(f'For given weights (w1={w1}, w2={w2}): Valid -w0 = {minus_w0}')
        print('Verification (w1*x1 + w2*x2 >= -w0 -> 1 else 0):')
        for (x1, x2), y in zip(inputs, truth_table):
            out = 1 if (w1*x1 + w2*x2 >= minus_w0) else 0
            print(f'  Input: ({x1}, {x2}) -> Output: {out} (Expected: {y})')
        return minus_w0
    else:
        print(f'For given weights (w1={w1}, w2={w2}): No valid -w0 exists.')
        return None

truth_tables = {
    'AND': [0, 0, 0, 1],
    'OR': [0, 1, 1, 1],
    'NOR': [1, 0, 0, 0]
}

print('== Implementing AND ==')
get_minus_w0(w1=1, w2=1, gate_name='AND', truth_table=truth_tables['AND'])
print('\n== Implementing OR ==')
get_minus_w0(w1=1, w2=1, gate_name='OR', truth_table=truth_tables['OR'])
print('\n== Implementing NOR ==')
get_minus_w0(w1=-1, w2=-1, gate_name='NOR', truth_table=truth_tables['NOR'])
```
